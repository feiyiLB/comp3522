```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
COMMUTE_CHOICE=pd.read_csv("../Data/Factors/18_COMMUTE_CHOICE.csv")
COMMUTE_CHOICE
```

```python
COMMUTE_CHOICE_SELECTED = pd.concat([COMMUTE_CHOICE.iloc[[4]], COMMUTE_CHOICE.iloc[6:120]], ignore_index=True)
COMMUTE_CHOICE_SELECTED.iloc[0,1]='工作地点'

COMMUTE_CHOICE_SELECTED.columns=COMMUTE_CHOICE_SELECTED.iloc[0]
COMMUTE_CHOICE_SELECTED=COMMUTE_CHOICE_SELECTED.iloc[1:]
COMMUTE_CHOICE_SELECTED
```

```python
COMMUTE_CHOICE_SELECTED_INTERNAL=COMMUTE_CHOICE_SELECTED[COMMUTE_CHOICE_SELECTED['工作地点']=='同區工作 (4)']
COMMUTE_CHOICE_SELECTED_INTERNAL=COMMUTE_CHOICE_SELECTED_INTERNAL.iloc[:, list(range(4)) + list(range(6,8))]
COMMUTE_CHOICE_SELECTED_INTERNAL
```

```python
fig, ax = plt.subplots(figsize=(12, 6))

# 交通方式类别
categories = ["香港鐵路（本地線） (1)", "巴士 (2)", "	公共小巴 (3)", "公司巴士／小巴"]
bottom = None  # 用于堆叠
c
# 绘制堆积柱形图
for category in categories:
    ax.bar(COMMUTE_CHOICE_SELECTED_INTERNAL["前赴工作地點的主要交通方式"], 
           COMMUTE_CHOICE_SELECTED_INTERNAL[category], 
           label=category, bottom=bottom)
    bottom = COMMUTE_CHOICE_SELECTED_INTERNAL[categories[:categories.index(category)+1]].sum(axis=1)

# 设置标签和标题
ax.set_xlabel("工作地点")
ax.set_ylabel("人数")
ax.set_title("各区主要交通方式人数堆积条形图")
ax.legend()

# 旋转 x 轴标签以避免重叠
plt.xticks(rotation=45)

# 显示图表
plt.show()
```
