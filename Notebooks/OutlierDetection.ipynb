```python
import pandas as pd
import matplotlib.pyplot as plt


data = pd.read_excel("../Data/BUS/Bus_FE_Processed.xlsx",engine='openpyxl')


# 确保PRICE列存在
if 'PRICE' in data.columns:
    # 绘制直方图
    plt.figure(figsize=(10, 6))
    plt.hist(data['PRICE'], bins=50, color='skyblue', edgecolor='black')
    
    # 添加标题和标签
    plt.title('Histogram of PRICE', fontsize=15)
    plt.xlabel('PRICE', fontsize=12)
    plt.ylabel('Frequency', fontsize=12)
    
    # 显示网格
    plt.grid(axis='y', alpha=0.75)
    
    # 显示图形
    plt.show()
else:
    print("Error: 'PRICE' column not found in the Excel file.")
    print("Available columns are:", list(data.columns))
```
